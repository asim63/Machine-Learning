As you can see, the age is float making your memory usage higher, well for a small dataset it wont make much difference, 
but in a larger scale, its better to save memory. So you can convert it into integer.

Oopsies, found out that age actually has some decimal values too, so i couldnt use ```.astype(int)``` or `.astype('Int64')`, so need to round that as well.

Apparently the size increased, LOL you will figure out in future. 

````python
df['Age'].describe()
# print(df['Age'][df['Age'] % 1 != 0])
df.loc[
    df['Age'].notna() & (df['Age'] % 1 != 0)
]
````

````python
df['Age'] = df['Age'].round().astype('Int64')
````
