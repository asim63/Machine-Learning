3. Opening a csv file from an URL ( use if csv file has to be loaded from server )

````python
import io
import pandas as pd
import requests

url = 'https://example.com'
headers = {
    'User-Agent': 'Mozilla/5.0'
}
req = requests.get(url, headers=headers)
df = pd.read_csv(io.StringIO(req.text))
````

6. Header Parameter (first row data navayi, column name thyo vane, to make the first row as column names)

````python
pd.read_csv("../datasets/customer_churn_dataset.csv", header=1)
````

7. use_cols parameter

````python
#suppose kunai selected features matra chaiyeko xa
pd.read_csv("../datasets/customer_churn_dataset.csv",usecols = ['Age','Gender'])
````

8. Skiprows/nrows Parameter

````python
pd.read_csv("../datasets/customer_churn_dataset.csv", skiprows=[1,3])

#nrows ( jati ota import garne ho rows teti matra aauxa)

pd.read_csv("../datasets/customer_churn_dataset.csv", nrows = 100)
````

9. Encoding parameter (encoding utf-8 bahek aru kei vayeko bela)

````python
pd.read_csv("../datasets/customer_churn_dataset.csv", encoding= '')
````

10. Skip bad lines

````python
pd.read_csv("../datasets/customer_churn_dataset.csv", on_bad_lines = "skip")
````

11. dtypes parameter (target is float, kina waste garne space, so integer ma change garna milxa)

````python
pd.read_csv("../datasets/customer_churn_dataset.csv", dtype={'Churn':'Int64'})

#removing the row with NaN value
#
#k rahexa vanda kheri, if you want to preserve the missing (Nan) values, use Int64 as shown;
#else chai use: dtype = {'Churn':int}
````

````python
#Customer Id, Age;  lai pani gardim nata
pd.read_csv("../datasets/customer_churn_dataset.csv", dtype={'CustomerID':'Int64', 'Age':'Int64'})
````

12. Handling Dates

````python
pd.read_csv("../datasets/customer_churn_dataset.csv", parse_dates=['date'])
#'date' chai feature ko name ho la
````

````python
def rename(name):
    if name == "Female":
        return "F"
    elif name =="Male": 
        return "M"
    else:
        return name
````

13. Convertors

````python
pd.read_csv("../datasets/customer_churn_dataset.csv", converters={'Gender': rename})
````

14. na_values parameter (NaN ko satta -, wa anything else huna ni sakxa kaile kai testo ma you will suru mei specify the kind of values which are NaN)

````python
pd.read_csv("../datasets/customer_churn_dataset.csv", na_values=['Male'])
#hyphen ( - ) harma use garne ni pheri Male ta example deko
````

15. Loading a huge dataset in chunks

````python
dfs = pd.read_csv("../datasets/customer_churn_dataset.csv", chunksize= 5000)
dfs
````

````python
for chunks in dfs:
    print(chunks.shape)
````

loc and iloc 

```.loc``` is used to select rows and columns using their labels
```py
df.loc[row_label, column_label]
```
```.iloc``` is used to select rows and columns using integer positions
```py
df.iloc[row_position, column_postion]
```
