Univariate analysis means the analysis of a single feature/column of the data. We will be using the titanic dataset to learn this.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../datasets/titanic_dataset.csv")
```

```python
print(df.shape)
df.sample(5)
```

1. Categorical Data

a. Countplot


```python
# sns.countplot(x = 'Embarked', data=df)
df['Pclass'].value_counts().plot(kind = 'bar')
```

b. Pie Chart

```python
df['Pclass'].value_counts().plot(kind = 'pie', autopct='%.2f')
```

2. Numerical Data

a. Histogram

```python
# plt.hist(df['Age'], bins = 30)
sns.histplot(x='Age', data=df, bins = 30)
```

b. Distplot

```python
# sns.displot(x='Age', data=df, kde=True)
sns.histplot(x='Age', data=df, kde=True, stat='density') # For PDF, probability aauxa side ma
```

c. Box Plot

```python
sns.boxplot(x= 'Fare', data= df)
```
