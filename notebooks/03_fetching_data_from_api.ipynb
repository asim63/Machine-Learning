```python
import pandas as pd
import requests
import os
from dotenv import load_dotenv

load_dotenv()
MY_API_KEY = os.getenv("TMDB_API_KEY")
```

```python
response = requests.get(f"https://api.themoviedb.org/3/movie/top_rated?api_key={MY_API_KEY}&page=1")
response.json()
```

```python
df = pd.DataFrame(response.json()['results'])
df = df[['id','title','vote_average','vote_count','popularity','release_date']]
df
```

```python
df = pd.DataFrame(response.json()['results'])[['id','title','vote_average','vote_count','popularity','release_date']]
df
```

```python
df = pd.DataFrame()
dfs=[]
for i in range(1,501):
    response = requests.get(f"https://api.themoviedb.org/3/movie/top_rated?api_key={MY_API_KEY}&page={i}")
    
    if response.status_code != 200:
        print(f"Request failed on page {i}")
        print(response.status_code)
        print(response.json())
        break
    
    temp_df= pd.DataFrame(response.json()['results'])[['id','title','vote_average','vote_count','popularity','release_date']]
    dfs.append(temp_df)
    
df = pd.concat(dfs, ignore_index=True)
print(df.head())
print(df.shape)
```

```python
df.to_csv("../datasets/movies.csv")
```
