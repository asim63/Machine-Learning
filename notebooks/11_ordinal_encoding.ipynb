Okay, so Ordinal Encoding is done on ordinal data.
Now , what is ordinal data?
- In which there is clear hierarchy or order of the data. eg: Grades in exam, A is greater than B and so on.


Nominal Encoding is done on normal data, like states, you dont have the order or any hierarchy for states do you?


Also you need to know about label encoding, label encoding is done when the output label is categorical as well, else do ordinal encoding


```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../datasets/customer.csv")
df.sample(5)
```

```python
df = df.iloc[:,2:] #Gender lai nakelo, cause its not ordinal, its nominal (Nominal ma One hot encoding laagxa we will do later)
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,0:2],
                                                    df.iloc[:,-1],
                                                    test_size=0.2,
                                                    random_state=42)
```

```python
X_train
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
```

```python
oe.fit_transform(X_train)
```

```python
oe.categories_
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```
