```python
rating_map = {
    'One':1,
    'Two':2,
    'Three':3,
    'Four':4,
    'Five':5
}
```

```python

name = []
price = []
availability = []
rating = []
for j in range (1, 51):
    url=f"https://books.toscrape.com/catalogue/page-{j}.html"
    headers = {'User-Agent': 'Mozilla/5.0'}
    response = requests.get(url,headers=headers)
    # print(response.status_code)
    webpage = response.text
    
    soup = BeautifulSoup(webpage, 'lxml')
    books = soup.find_all('li', class_= 'col-xs-6 col-sm-4 col-md-3 col-lg-3')
    # len(books)

    for i in books:
        name.append(i.find('h3').text.strip())
        price.append(i.find('p', class_='price_color').text.strip().replace("Â", ""))
        availability.append(i.find('p', class_='instock').text.strip())
        rat_num = i.find('p',class_='star-rating')
        rating.append(rating_map[rat_num['class'][1]])
        
        
    
d ={
    'name':name,
    'price':price,
    'availability':availability,
    'rating':rating
}
df = pd.DataFrame(d)
df
        
```
